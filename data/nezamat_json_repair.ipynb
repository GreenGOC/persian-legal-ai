```python
from pathlib import Path
```

```python
def repair_json_file(input_path, index, id_line):
    input_path = Path(input_path)
    output_path = Path(f"valid_json/{index}.json")

    lines = input_path.read_text(encoding="utf-8").splitlines()
    output_lines = []
    brace_depth = 0
    bracket_depth = 0
    output_lines.append("{")
    if id_line:
        output_lines.append(f"\"id\": \"{id_line}\",")
    else:
        output_lines.append("\"id\": null,")
    in_string = False
    escaped = False
    for line in lines[1:-1]:
        line = line.strip()
        if not line:
            output_lines.append("")
            continue
        if line == '"children": [':
            line = '"children": []\n},'
        if line == '"structure": {':
            line = '"structure": [\n{'
            bracket_depth-=1
        cleaned = []
        i = 0
        while i < len(line):
            char = line[i]
            if in_string:
                cleaned.append(char)
                
                if escaped:
                    escaped = False
                elif char == "\\":
                    escaped = True
                elif char == '"':
                    in_string = False
                    
                i += 1
                continue
            
            if char == '"':
                in_string = True
                cleaned.append(char)
                
            elif char == "{":
                brace_depth += 1
                cleaned.append(char)

            elif char == "[":
                bracket_depth += 1
                cleaned.append(char)

            elif char == "}":
                if brace_depth > 0:
                    brace_depth -= 1
                    cleaned.append(char)
                else:
                    pass

            elif char == "]":
                if bracket_depth > 0:
                    bracket_depth -= 1
                    cleaned.append(char)
                else:
                    pass

            else:
                cleaned.append(char)

            i += 1
        output_lines.append("".join(cleaned))

    output_lines.append("]}")
    output_path.write_text(
        "\n".join(output_lines) + "\n",
        encoding="utf-8"
    )

    return output_path
```

```python
with open("./links/nezamat.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()
    for i in range(9, 41410):
        related_lines = lines[i*10:(i+1)*10]
        input_file = Path(f"./json/{i}.json")
        if not input_file.exists():
            continue
        output_file = repair_json_file(input_file, i, related_lines[4].replace("number---", "").strip())
        print(f"saved: {output_file}")
```
